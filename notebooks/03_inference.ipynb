```python
import pandas as pd
import joblib
```

```python
model = joblib.load(
    "../data/processed/random_forest_model.pkl",
    mmap_mode="r"
)

columns = joblib.load(
    "../data/processed/model_columns.pkl"
)

print("Modelo carregado!")
```

```python
new_data = pd.DataFrame({
    "Store": [1],
    "Dept": [1],
    "IsHoliday": [False],
    "Year": [2012],
    "Month": [11],
    "Day": [23],
    "Quarter": [4],
    "Temperature": [60],
    "Fuel_Price": [3.2],
    "MarkDown1": [5000],
    "MarkDown2": [0],
    "MarkDown3": [0],
    "MarkDown4": [0],
    "MarkDown5": [1500],
    "CPI": [220],
    "Unemployment": [7.5],
    "Size": [151315],
    "Week": [47],
    "DayOfWeek": [4],
    "IsWeekend": [False],
    "Type_B": [0],
    "Type_C": [0]
})
```

```python
new_data = new_data[columns]
```

```python
prediction = model.predict(new_data)

print(f"Venda prevista: ${prediction[0]:,.2f}")
```

```python
normal_scenario = pd.DataFrame({
    "Store": [1],
    "Dept": [1],
    "IsHoliday": [False],
    "Year": [2012],
    "Month": [11],
    "Day": [23],
    "Quarter": [4],
    "Temperature": [60],
    "Fuel_Price": [3.2],
    "MarkDown1": [5000],
    "MarkDown2": [0],
    "MarkDown3": [0],
    "MarkDown4": [0],
    "MarkDown5": [1500],
    "CPI": [220],
    "Unemployment": [7.5],
    "Size": [151315],
    "Week": [47],
    "DayOfWeek": [4],
    "IsWeekend": [False],
    "Type_B": [0],
    "Type_C": [0]
})

normal_scenario = normal_scenario.reindex(
    columns=columns,
    fill_value=0
)

normal_prediction = model.predict(normal_scenario)[0]
```

```python
promotion_scenario = normal_scenario.copy()

promotion_scenario.loc[0, "IsHoliday"] = True
promotion_scenario.loc[0, "Month"] = 12
promotion_scenario.loc[0, "Quarter"] = 4
promotion_scenario.loc[0, "MarkDown1"] = 25000
promotion_scenario.loc[0, "MarkDown5"] = 8000

promotion_prediction = model.predict(promotion_scenario)[0]
```

```python
results = pd.DataFrame({
    "Cenário": ["Semana normal", "Semana promocional"],
    "Venda prevista": [
        normal_prediction,
        promotion_prediction
    ]
})

results
```

```python
difference = promotion_prediction - normal_prediction

print(f"Previsão normal: ${normal_prediction:,.2f}")
print(f"Previsão promocional: ${promotion_prediction:,.2f}")
print(f"Diferença: ${difference:,.2f}")
```

```python
def predict_weekly_sales(input_data: dict) -> float:
    """
    Recebe os dados de uma loja/departamento e retorna
    a previsão de vendas semanais.
    """
    prediction_data = pd.DataFrame([input_data])

    prediction_data = prediction_data.reindex(
        columns=columns,
        fill_value=0
    )

    prediction = model.predict(prediction_data)

    return float(prediction[0])
```

```python
sample_input = {
    "Store": 1,
    "Dept": 1,
    "IsHoliday": False,
    "Year": 2012,
    "Month": 11,
    "Day": 23,
    "Quarter": 4,
    "Temperature": 60,
    "Fuel_Price": 3.2,
    "MarkDown1": 5000,
    "MarkDown2": 0,
    "MarkDown3": 0,
    "MarkDown4": 0,
    "MarkDown5": 1500,
    "CPI": 220,
    "Unemployment": 7.5,
    "Size": 151315,
    "Week": 47,
    "DayOfWeek": 4,
    "IsWeekend": False,
    "Type_B": 0,
    "Type_C": 0
}

predicted_sales = predict_weekly_sales(sample_input)

print(f"Venda prevista: ${predicted_sales:,.2f}")
```
