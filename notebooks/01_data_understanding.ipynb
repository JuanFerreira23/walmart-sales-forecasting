```python
features[[
    "MarkDown1",
    "MarkDown2",
    "MarkDown3",
    "MarkDown4",
    "MarkDown5"
]].head(15)
```

```python
markdown_cols = [
    "MarkDown1",
    "MarkDown2",
    "MarkDown3",
    "MarkDown4",
    "MarkDown5"
]

features[markdown_cols] = features[markdown_cols].fillna(0)
```

```python
features[markdown_cols].isnull().sum()
```

```python
features.isnull().sum()
```

```python
features["CPI"] = features["CPI"].fillna(features["CPI"].median())

features["Unemployment"] = (
    features["Unemployment"]
    .fillna(features["Unemployment"].median())
)
```

```python
features.isnull().sum()
```

```python
# Junta train + features
data = train.merge(
    features,
    on=["Store", "Date", "IsHoliday"],
    how="left"
)

# Junta o resultado com stores
data = data.merge(
    stores,
    on="Store",
    how="left"
)
```
